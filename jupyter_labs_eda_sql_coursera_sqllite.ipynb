## Tasks

Now write and execute SQL queries to solve the assignment tasks.

**Note: If the column names are in mixed case enclose it in double quotes
   For Example "Landing_Outcome"**

### Task 1




##### Display the names of the unique launch sites  in the space mission


```python
# Task 1: Display unique launch site names
query = """
SELECT DISTINCT "Launch_Site"
FROM SPACEXTBL;
"""

# Execute the query
cur.execute(query)
results = cur.fetchall()

# Print the unique launch site names
print("Unique Launch Sites:")
for row in results:
    print(row[0])

```


### Task 2


#####  Display 5 records where launch sites begin with the string 'CCA'


```python
query2 = """
SELECT *
FROM SPACEXTBL
WHERE Launch_Site LIKE 'CCA%'
LIMIT 5;
"""

cur.execute(query2)
results2 = cur.fetchall()

print("\n5 Records where Launch Site starts with 'CCA':")
for row in results2:
    print(row)
```

### Task 3




##### Display the total payload mass carried by boosters launched by NASA (CRS)


```python
query3 = """
SELECT SUM(PAYLOAD_MASS__KG_)
FROM SPACEXTBL
WHERE Customer = 'NASA (CRS)';
"""

cur.execute(query3)
results3 = cur.fetchall()
print("\nTotal payload mass carried by boosters launched by NASA (CRS):")
print(results3[0][0])
```

### Task 4




##### Display average payload mass carried by booster version F9 v1.1


```python
query4 = """
SELECT AVG(PAYLOAD_MASS__KG_)
FROM SPACEXTBL
WHERE Booster_Version = 'F9 v1.1';
"""

cur.execute(query4)
results4 = cur.fetchall()
print("\nAverage payload mass carried by booster version F9 v1.1:")
print(results4[0][0])
```

### Task 5

##### List the date when the first succesful landing outcome in ground pad was acheived.


_Hint:Use min function_


```python
query5 = """
SELECT MIN(Date)
FROM SPACEXTBL
WHERE Landing_Outcome LIKE 'Success (ground pad)';
"""

cur.execute(query5)
results5 = cur.fetchall()
print("\nDate when the first succesful landing outcome in ground pad was acheived:")
print(results5[0][0])
```

### Task 6

##### List the names of the boosters which have success in drone ship and have payload mass greater than 4000 but less than 6000


```python
query6 = """
SELECT Booster_Version
FROM SPACEXTBL
WHERE Landing_Outcome = 'Success (drone ship)' AND PAYLOAD_MASS__KG_ BETWEEN 4000 AND 6000;
"""

cur.execute(query6)
results6 = cur.fetchall()
print("\nBoosters which have success in drone ship and payload mass between 4000 and 6000:")
for row in results6:
    print(row[0])
```

### Task 7




##### List the total number of successful and failure mission outcomes


```python
query7 = """
SELECT Mission_Outcome, COUNT(*) AS TotalCount
FROM SPACEXTBL
GROUP BY Mission_Outcome;
"""

cur.execute(query7)
results7 = cur.fetchall()
print("\nTotal number of successful and failure mission outcomes:")
for row in results7:
    print(f"{row[0]}: {row[1]}")
```

### Task 8



##### List the   names of the booster_versions which have carried the maximum payload mass. Use a subquery


```python
query8 = """
SELECT Booster_Version
FROM SPACEXTBL
WHERE PAYLOAD_MASS__KG_ = (SELECT MAX(PAYLOAD_MASS__KG_) FROM SPACEXTBL);
"""

cur.execute(query8)
results8 = cur.fetchall()
print("\nBooster versions which have carried the maximum payload mass:")
for row in results8:
    print(row[0])
```

### Task 9


##### List the records which will display the month names, failure landing_outcomes in drone ship ,booster versions, launch_site for the months in year 2015.

**Note: SQLLite does not support monthnames. So you need to use  substr(Date, 6,2) as month to get the months and substr(Date,0,5)='2015' for year.**


```python
query9 = """
SELECT Landing_Outcome
FROM SPACEXTBL
WHERE Landing_Outcome LIKE 'Failure (drone ship)'
ORDER BY Date DESC;
"""

cur.execute(query9)
results9 = cur.fetchall()
print("\nFailed landing outcomes in drone ship (descending order by Date):")
for row in results9:
    print(row[0])
```

### Task 10




##### Rank the count of landing outcomes (such as Failure (drone ship) or Success (ground pad)) between the date 2010-06-04 and 2017-03-20, in descending order.


```python
query10 = """
SELECT Booster_Version
FROM SPACEXTBL
WHERE Landing_Outcome LIKE 'Success (ground pad)' OR Landing_Outcome LIKE 'Success (drone ship)';
"""

cur.execute(query10)
results10 = cur.fetchall()
print("\nBooster names which have done successful self landing:")
for row in results10:
    print(row[0])

# Close the connection
con.close()
```
